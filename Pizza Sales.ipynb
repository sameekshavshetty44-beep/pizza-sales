```python
#Bar Chart – Revenue by Category
#Requirement: Display revenue generated by each pizza category.

revenue = df.groupby('pizza_category')['total_price'].sum()

x = revenue.index
y = revenue.values

plt.figure(figsize=(6,4))
plt.bar(x, y)

plt.title("Revenue by Pizza Category")
plt.xlabel("Pizza Category")
plt.ylabel("Revenue")

plt.show()
```

```python
#Horizontal Bar Chart – Top 10 Pizzas by Sales
#Requirement: Display top 10 best-selling pizzas.

top10 = df.groupby('pizza_name')['quantity'].sum().sort_values(ascending=False).head(10)

x = top10.values
y = top10.index

plt.figure(figsize=(8,5))
plt.barh(y, x)

plt.title("Top 10 Best Selling Pizzas")
plt.xlabel("Quantity Sold")
plt.ylabel("Pizza Name")

plt.show()
```

```python
#Pie Chart – Revenue Contribution by Category
#Requirement: Show revenue contribution of each category.

revenue = df.groupby('pizza_category')['total_price'].sum()

labels = revenue.index
values = revenue.values

plt.figure(figsize=(6,6))
plt.pie(values,
        labels=labels,
        autopct='%1.2f%%')

plt.title("Revenue Contribution by Category")

plt.show()
```

```python
#Line Chart – Hour-wise Sales Trend
#Requirement:Display quantity sold during each hour.

sales = df.groupby('hour')['quantity'].sum()

x = sales.index
y = sales.values

plt.figure(figsize=(8,4))

plt.plot(x, y, marker='o')

plt.title("Hour-wise Sales Trend")
plt.xlabel("Hour")
plt.ylabel("Quantity Sold")

plt.show()
```

```python
#Find Hour-wise Revenue Trend

revenue_trend = df.groupby('hour')['total_price'].sum()
print(revenue_trend)

x = revenue_trend.index
y = revenue_trend.values

plt.figure(figsize=(8,4))

plt.plot(x, y, marker='o')

plt.title("Hour-wise Revenue Trend")
plt.xlabel("Hour")
plt.ylabel("Revenue")

plt.show()
```
